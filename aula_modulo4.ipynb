```python
# livros que possuem mais comentários do que o número de reviews
df[df.ratings_count < df.text_reviews_count]
```

```python
# livros que possuem  mais número de reviews do que o número de comentários
df[df.ratings_count > df.text_reviews_count]
```

```python
# Autores mais influentes
autores = df.groupby('authors').agg({'average_rating':'mean'}).sort_values(by='average_rating',ascending=False)
autores.head(10)
```

```python
# Autores mais influentes agrupados por autor, quantidade de livros e média de avaliação

autores = df.groupby('authors').agg(
    total_livros = ('title','count'),
    media_avaliacao = ('average_rating','mean')
).sort_values(by='total_livros', ascending= False)
autores.head(10)
```

```python
# Analise por idioma:
idiomas = df.groupby('language_code').agg(
    total_livros = ('title','count'),
    media_avaliacao = ('average_rating','mean')
).sort_values(by='total_livros', ascending= False)
idiomas.head(10)
```

# Relatório:

O conjunto de dados `books.csv` contém informações sobre uma variedade de livros. As colunas incluem `bookID`, `title`, `authors`, `average_rating`, `isbn`, `isbn13`, `language_code`, `num_pages`, `ratings_count`, `text_reviews_count`, `publication_date` e `publisher`.

## Análise dos Dados

### Carregamento e Inspeção Inicial

O dataset foi carregado a partir do arquivo `books.csv`. Uma inspeção inicial (`df.info()`) revelou que o dataset possui **11.123 entradas** e **12 colunas**. Todas as colunas possuem valores não nulos, indicando que não há dados faltantes. Os tipos de dados incluem inteiros (`int64`), objetos (`object`) e ponto flutuante (`float64`).

### Estatísticas Descritivas

Abaixo estão as estatísticas descritivas para as colunas numéricas do dataset, como `bookID`, `average_rating`, `isbn13`, `num_pages`, `ratings_count` e `text_reviews_count`.

```python
display(df.describe())
```

Principais observações das estatísticas descritivas:

*   **`average_rating`**: A classificação média varia de 0.0 a 5.0, com uma média de aproximadamente **3.93**. A moda (valor mais frequente) é **4.0** e a mediana é **3.96**. A variância é de **0.12** e o desvio padrão é de **0.35**, indicando que as classificações não estão amplamente dispersas.
*   **`num_pages`**: O número de páginas varia bastante, de 0 a 6576, com uma média de cerca de **336 páginas**.
*   **`ratings_count`**: O número de avaliações é muito variável, com alguns livros tendo zero avaliações e outros com milhões, indicando grande popularidade de certos títulos. A média é de aproximadamente 17.942 avaliações.
*   **`text_reviews_count`**: Similarmente, a contagem de resenhas de texto também varia significativamente, com uma média de 542 resenhas.

### Análise de Autores

Foi realizada uma análise para identificar os autores mais influentes, considerando o número total de livros e a média de suas avaliações. Os 10 autores com mais livros no dataset são:

```python
display(autores.head(10))
```

Podemos observar que **P.G. Wodehouse** e **Stephen King** lideram em número de livros (40 cada), com classificações médias respeitáveis.

### Análise por Idioma

A distribuição de livros e suas classificações médias também foi investigada por `language_code`. Os 10 idiomas com mais livros no dataset são:

```python
display(idiomas.head(10))
```

O idioma **inglês (`eng`)** é predominante, representando a vasta maioria dos livros, seguido por `en-US` e `spa` (espanhol). As classificações médias são consistentemente altas entre os idiomas mais comuns.

### Discrepâncias em Avaliações e Comentários

Foi identificado um pequeno número de livros onde o `ratings_count` (contagem de avaliações) é menor que o `text_reviews_count` (contagem de resenhas de texto). Isso pode indicar anomalias nos dados ou casos específicos onde resenhas foram registradas sem uma avaliação numérica correspondente. Por outro lado, a vasta maioria dos livros segue a lógica esperada, com mais avaliações do que resenhas.
